```python
import requests
import json
from datetime import datetime

try:
    token = notebookutils.credentials.getToken("pbi")
    print("✅ Token obtenido via notebookutils")
except Exception:
    try:
        token = mssparkutils.credentials.getToken("pbi")
        print("✅ Token obtenido via mssparkutils")
    except Exception as e:
        raise RuntimeError(f"No se pudo obtener el token. Ejecuta dentro de Fabric: {e}")

HEADERS = {
    "Authorization": f"Bearer {token}",
    "Content-Type": "application/json"
}
FABRIC_API  = "https://api.fabric.microsoft.com/v1"
POWERBI_API = "https://api.powerbi.com/v1.0/myorg"
print("🔗 APIs configuradas")

def get_capacity_info(capacity_id):
    """
    Obtiene info de la capacidad Fabric.
    Estrategia:
      1. Busca capacity_id en la lista de capacidades visibles (API)
      2. Si no la encuentra (sin permisos), usa CAPACITY_INFO_OVERRIDE
    Retorna dict con: displayName, sku, region, state, admins, source
    """
    # Intento 1: Fabric API /capacities
    try:
        caps_fabric = api_get(f"{FABRIC_API}/capacities", "capacities_fabric")
        for cap in (caps_fabric if isinstance(caps_fabric, list) else []):
            if cap.get("id") == capacity_id:
                print(f"  ✅ Capacidad encontrada via Fabric API: {cap.get('displayName')} [{cap.get('sku')}]")
                return {
                    "displayName": cap.get("displayName", "—"),
                    "sku":         cap.get("sku", "—"),
                    "region":      cap.get("region", "—"),
                    "state":       cap.get("state", "—"),
                    "admins":      [],
                    "source":      "API"
                }
    except Exception as e:
        print(f"  ⚠️ Fabric API capacities: {e}")
 
    # Intento 2: Power BI API /capacities (tiene admins y users)
    try:
        caps_pbi = api_get(f"{POWERBI_API}/capacities", "capacities_pbi")
        for cap in (caps_pbi if isinstance(caps_pbi, list) else []):
            if cap.get("id") == capacity_id:
                admins = cap.get("admins", [])
                print(f"  ✅ Capacidad encontrada via PBI API: {cap.get('displayName')} [{cap.get('sku')}]")
                return {
                    "displayName": cap.get("displayName", "—"),
                    "sku":         cap.get("sku", "—"),
                    "region":      cap.get("region", "—"),
                    "state":       cap.get("state", "—"),
                    "admins":      admins,
                    "source":      "API"
                }
    except Exception as e:
        print(f"  ⚠️ PBI API capacities: {e}")
 
    # Fallback: usar override manual de Celda 2b
    print(f"  ℹ️  Capacidad {capacity_id[:8]}… no visible via API — usando CAPACITY_INFO_OVERRIDE")
    return {
        "displayName": CAPACITY_INFO_OVERRIDE.get("displayName", "—"),
        "sku":         CAPACITY_INFO_OVERRIDE.get("sku", "—"),
        "region":      CAPACITY_INFO_OVERRIDE.get("region", "—"),
        "state":       CAPACITY_INFO_OVERRIDE.get("state", "Active"),
        "admins":      CAPACITY_INFO_OVERRIDE.get("admins", []),
        "source":      "manual"
    }
 
print("✅ get_capacity_info() definida")
```

```python
import io, base64
import tempfile, os as _os
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import cm
from reportlab.lib import colors
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, PageBreak,
    Table as RLTable, TableStyle, HRFlowable, Image as RLImage
)
from reportlab.lib.enums import TA_CENTER, TA_LEFT, TA_RIGHT

# Guardar logo en disco temporal para que ReportLab lo lea correctamente
_LOGO_TMP = _os.path.join(tempfile.gettempdir(), "project_logo.jpg")
with open(_LOGO_TMP, "wb") as _f:
    _f.write(LOGO_BYTES)
print(f"✅ Logo guardado en {_LOGO_TMP}")

# Paleta FabricDoc
RL_ACCENT    = colors.HexColor("#0078D4")
RL_GRAY_DARK  = colors.HexColor("#404040")
RL_GRAY_MID   = colors.HexColor("#808080")
RL_PINK_LIGHT = colors.HexColor("#F0F6FF")
RL_WHITE      = colors.white
RL_BORDER     = colors.HexColor("#E0E0E0")

PAGE_W, PAGE_H = A4
MARGIN = 2.5 * cm
CONTENT_W = PAGE_W - 2 * MARGIN


def pdf_styles():
    S = {}
    S["cover_title"] = ParagraphStyle("CT",
        fontSize=34, leading=42, textColor=RL_GRAY_DARK,
        fontName="Helvetica-Bold", alignment=TA_LEFT)
    S["cover_sub"] = ParagraphStyle("CS",
        fontSize=12, leading=18, textColor=RL_GRAY_MID,
        fontName="Helvetica", alignment=TA_LEFT)
    S["h1"] = ParagraphStyle("H1",
        fontSize=13, leading=18, textColor=RL_GRAY_DARK,
        fontName="Helvetica-Bold", spaceAfter=6, spaceBefore=14,
        borderPadding=(0,0,4,0))
    S["h2"] = ParagraphStyle("H2",
        fontSize=10, leading=14, textColor=RL_ACCENT,
        fontName="Helvetica-Bold", spaceAfter=4, spaceBefore=10)
    S["h3"] = ParagraphStyle("H3",
        fontSize=9, leading=13, textColor=RL_GRAY_DARK,
        fontName="Helvetica-BoldOblique", spaceAfter=3, spaceBefore=6)
    S["body"] = ParagraphStyle("Body",
        fontSize=9, leading=13, textColor=RL_GRAY_DARK,
        fontName="Helvetica", spaceAfter=4)
    S["code"] = ParagraphStyle("Code",
        fontSize=7, leading=9, textColor=colors.black,
        fontName="Courier", spaceAfter=4,
        backColor=colors.HexColor("#F5F5F5"),
        leftIndent=8, rightIndent=8)
    S["italic"] = ParagraphStyle("Italic",
        fontSize=9, leading=13, textColor=RL_GRAY_MID,
        fontName="Helvetica-Oblique")
    S["notice"] = ParagraphStyle("Notice",
        fontSize=7.5, leading=11, textColor=RL_GRAY_MID,
        fontName="Helvetica-Oblique")
    S["backcover"] = ParagraphStyle("BC",
        fontSize=13, leading=18, textColor=RL_ACCENT,
        fontName="Helvetica-Bold", alignment=TA_CENTER)
    return S


def rl_meta_table(items_dict):
    data = [["Propiedad", "Valor"]] + [[k, str(v) if v else "—"] for k, v in items_dict.items()]
    col_w = [4 * cm, CONTENT_W - 4 * cm]
    style = TableStyle([
        ("BACKGROUND",    (0,0), (-1,0), RL_ACCENT),
        ("TEXTCOLOR",     (0,0), (-1,0), RL_WHITE),
        ("FONTNAME",      (0,0), (-1,0), "Helvetica-Bold"),
        ("FONTSIZE",      (0,0), (-1,-1), 8.5),
        ("BACKGROUND",    (0,1), (0,-1), RL_PINK_LIGHT),
        ("FONTNAME",      (0,1), (0,-1), "Helvetica-Bold"),
        ("ROWBACKGROUNDS",(1,1), (-1,-1), [RL_WHITE, RL_PINK_LIGHT]),
        ("GRID",          (0,0), (-1,-1), 0.3, RL_BORDER),
        ("VALIGN",        (0,0), (-1,-1), "MIDDLE"),
        ("LEFTPADDING",   (0,0), (-1,-1), 6),
        ("RIGHTPADDING",  (0,0), (-1,-1), 6),
        ("TOPPADDING",    (0,0), (-1,-1), 4),
        ("BOTTOMPADDING", (0,0), (-1,-1), 4),
    ])
    t = RLTable(data, colWidths=col_w, repeatRows=1)
    t.setStyle(style)
    return t


def rl_schema_table(headers, rows, col_w=None):
    n = len(headers)
    if not col_w:
        col_w = [CONTENT_W / n] * n
    data = [headers] + rows
    style = TableStyle([
        ("BACKGROUND",    (0,0), (-1,0), RL_ACCENT),
        ("TEXTCOLOR",     (0,0), (-1,0), RL_WHITE),
        ("FONTNAME",      (0,0), (-1,0), "Helvetica-Bold"),
        ("FONTSIZE",      (0,0), (-1,-1), 8),
        ("ROWBACKGROUNDS",(0,1), (-1,-1), [RL_WHITE, RL_PINK_LIGHT]),
        ("GRID",          (0,0), (-1,-1), 0.3, RL_BORDER),
        ("VALIGN",        (0,0), (-1,-1), "MIDDLE"),
        ("LEFTPADDING",   (0,0), (-1,-1), 5),
        ("RIGHTPADDING",  (0,0), (-1,-1), 5),
        ("TOPPADDING",    (0,0), (-1,-1), 3),
        ("BOTTOMPADDING", (0,0), (-1,-1), 3),
    ])
    t = RLTable(data, colWidths=col_w, repeatRows=1)
    t.setStyle(style)
    return t


def pdf_h1_with_rule(text):
    """Retorna lista de flowables: H1 + línea magenta."""
    S = pdf_styles()
    return [
        Paragraph(text.upper(), S["h1"]),
        HRFlowable(width="100%", thickness=2, color=RL_ACCENT, spaceAfter=6)
    ]


def build_pdf_document(data, pdf_path):
    S  = pdf_styles()
    doc = SimpleDocTemplate(
        pdf_path, pagesize=A4,
        leftMargin=MARGIN, rightMargin=MARGIN,
        topMargin=MARGIN,  bottomMargin=MARGIN
    )
    story = []

    # ── PORTADA ─────────────────────────────────────────────────
    logo_img = RLImage(_LOGO_TMP, width=4.5*cm, height=2.8*cm, kind="bound")
    story += [
        logo_img,
        Spacer(1, 4*cm),
        Paragraph("Documentación<br/>técnica", S["cover_title"]),
        Spacer(1, 0.5*cm),
        Paragraph(f"{DOC_CLIENT}<br/>Ref.: {DOC_REF}", S["cover_sub"]),
        Spacer(1, 0.3*cm),
        HRFlowable(width="20%", thickness=3, color=RL_ACCENT, hAlign="LEFT"),
        PageBreak()
    ]

    # ── INFO DOCUMENTO ──────────────────────────────────────────
    story += pdf_h1_with_rule("Información del documento")
    story.append(rl_meta_table({
        "Título":            "Documentación técnica",
        "Código":            DOC_REF,
        "Cliente":           DOC_CLIENT,
        "Nivel de seguridad":DOC_SECURITY,
        "Propietario":       DOC_AUTHOR
    }))
    story.append(Spacer(1, 0.4*cm))
    story.append(Paragraph("Registro de versiones", S["h2"]))
    story.append(rl_schema_table(
        ["Autor", "Descripción", "Versión", "Fecha"],
        [[DOC_AUTHOR, "Generado automáticamente por Fabric Documentator", DOC_VERSION, data["generated_at"][:10]]],
        col_w=[4*cm, 8*cm, 2*cm, 2.6*cm]
    ))
    story.append(Spacer(1, 0.4*cm))
    story.append(Paragraph(
        f"Aviso: Este documento es propiedad de {DOC_AUTHOR} y contiene información clasificada "
        f"según el nivel de seguridad definido ({DOC_SECURITY}), debiendo aplicarse las medidas de uso y custodia pertinentes.",
        S["notice"]
    ))
    story.append(PageBreak())

    # ── CONTENIDO POR WORKSPACE ─────────────────────────────────
    for ws in data["workspaces"]:

        # 1. OBJETO
        story += pdf_h1_with_rule("1. Objeto")
        story.append(Paragraph(
            f"La documentación post-despliegue del proyecto {DOC_CLIENT} ha sido generada automáticamente "
            f"a partir de los metadatos del workspace <b>{ws['name']}</b> con fecha {data['generated_at'][:10]}.",
            S["body"]
        ))
        story.append(Spacer(1, 0.4*cm))

        # 2. FABRIC WORKSPACE
        story += pdf_h1_with_rule("2. Fabric Workspace")
        story.append(Paragraph("2.1 Áreas de trabajo", S["h2"]))
        story.append(rl_meta_table({
            "Nombre":    ws["name"],
            "ID":        ws["id"],
            "Tipo":      ws.get("type","—"),
            "Capacidad": ws.get("capacityId","—"),
            "Estado":    ws.get("state","—")
        }))
        story.append(Spacer(1, 0.3*cm))

        # 3. CAPA DE DATOS
        story += pdf_h1_with_rule("3. Capa de datos")

        # Lakehouses
        story.append(Paragraph("3.1 Almacenamiento (Lakehouse)", S["h2"]))
        for lh in ws["lakehouses"]:
            story.append(Paragraph(f"Lakehouse: {lh['name']}", S["h3"]))
            story.append(rl_meta_table({"ID": lh.get("id",""), "SQL Endpoint": lh.get("sqlEndpoint","—")}))
            schemas_dict = {}
            for t in lh.get("tables", []):
                schemas_dict.setdefault(t.get("schema","default"), []).append(t)
            for sch, tbls in schemas_dict.items():
                story.append(Paragraph(f"Schema: {sch}", S["h3"]))
                story.append(rl_schema_table(
                    ["Tabla", "Formato", "Nº columnas"],
                    [[t.get("name",""), t.get("format","delta"), str(len(t.get("schema_cols",[])))] for t in tbls],
                    col_w=[9*cm, 4*cm, 3.6*cm]
                ))
                for t in tbls:
                    cols = t.get("schema_cols", [])
                    if cols:
                        story.append(Paragraph(f"{sch}.{t.get('name','')}", S["h3"]))
                        story.append(rl_schema_table(
                            ["Columna", "Tipo", "Comentario"],
                            [[c.get("column",""), c.get("type",""), c.get("comment","")] for c in cols],
                            col_w=[6*cm, 5*cm, 5.6*cm]
                        ))
            story.append(Spacer(1, 0.3*cm))

        # Data Warehouses
        story.append(Paragraph("3.2 Data Warehouses", S["h2"]))
        for wh in ws["warehouses"]:
            story.append(Paragraph(f"DW: {wh['name']}", S["h3"]))
            story.append(rl_meta_table({"ID": wh.get("id",""), "Conn": wh.get("connectionString","—")}))
            for t in wh.get("tables", []):
                story.append(Paragraph(f"{t.get('schema','')}.{t.get('name','')}", S["h3"]))
                cols = t.get("columns", [])
                if cols:
                    story.append(rl_schema_table(
                        ["Columna", "Tipo", "Nullable", "PK"],
                        [[c.get("column",""), c.get("type",""), "Sí" if c.get("nullable") else "No", "PK" if c.get("isPK") else ""] for c in cols],
                        col_w=[6*cm, 4.5*cm, 2.5*cm, 1.6*cm]
                    ))
        story.append(Spacer(1, 0.3*cm))

        # Semantic Models
        story.append(Paragraph("3.3 Semantic Models", S["h2"]))
        for m in ws["semantic_models"]:
            story.append(Paragraph(f"Modelo: {m.get('name','')}", S["h3"]))
            story.append(rl_meta_table({"ID": m.get("id",""), "Refreshable": "Sí" if m.get("isRefreshable") else "No"}))
            rels = m.get("relationships", [])
            if rels:
                story.append(rl_schema_table(
                    ["Tabla Origen", "Col Origen", "Tabla Destino", "Col Destino"],
                    [[r.get("fromTable",""), r.get("fromColumn",""), r.get("toTable",""), r.get("toColumn","")] for r in rels],
                    col_w=[4.2*cm, 3.5*cm, 4.2*cm, 3.5*cm]
                ))
        story.append(Spacer(1, 0.3*cm))

        # 4. ORQUESTACIÓN
        story += pdf_h1_with_rule("4. Capa de orquestación")

        story.append(Paragraph("4.2 Canalizaciones de datos (Pipelines)", S["h2"]))
        if ws["pipelines"]:
            story.append(rl_schema_table(
                ["Nombre", "Descripción", "Creado", "Modificado"],
                [
                    [
                        p.get("name", ""),
                        p.get("description", "") or "—",
                        (p.get("createdDate",  "") or "")[:10],
                        (p.get("modifiedDate", "") or "")[:10]
                    ]
                    for p in ws["pipelines"]
                ],
                col_w=[5.5*cm, 7*cm, 2.8*cm, 2.8*cm]
            ))
        else:
            story.append(Paragraph("Sin pipelines.", S["italic"]))

        story.append(Paragraph("4.3 Notebooks", S["h2"]))
        for nb in ws["notebooks"]:
            story.append(Paragraph(f"{nb.get('name','')}", S["h3"]))
            story.append(rl_meta_table({"ID": nb.get("id",""), "Creado": nb.get("createdDate","—")}))
            if INCLUDE_NOTEBOOK_CODE and nb.get("code"):
                for snippet in nb["code"]:
                    safe = snippet.replace("&","&amp;").replace("<","&lt;").replace(">","&gt;").replace("\n","<br/>").replace(" ","&nbsp;")
                    story.append(Paragraph(safe, S["code"]))

        story.append(Paragraph("Reports", S["h2"]))
        if ws["reports"]:
            story.append(rl_schema_table(
                ["Nombre", "Dataset", "Creado por", "Modificado"],
                [[r.get("name",""), r.get("datasetName",""), r.get("createdBy",""), (r.get("modifiedDateTime","") or "")[:10]] for r in ws["reports"]],
                col_w=[5*cm, 5*cm, 4*cm, 2.6*cm]
            ))
        else:
            story.append(Paragraph("Sin reports.", S["italic"]))

        story.append(PageBreak())

    # ── CONTRAPORTADA ────────────────────────────────────────────
    story += [
        Spacer(1, 8*cm),
        RLImage(_LOGO_TMP, width=6*cm, height=3.8*cm, kind="bound"),
        Spacer(1, 0.6*cm),
        Paragraph("github.com/ecartolariesteves", S["backcover"])
    ]

    doc.build(story)


print("✅ Funciones PDF (estilo FabricDoc) definidas")
```
